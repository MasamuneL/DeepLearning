```python
import os

import cv2

from ultralytics import YOLO

# Modelo de segmentación
model = YOLO("yolo26n-seg.pt")

# Video de entrada
cap = cv2.VideoCapture("/Users/vikingo/Documents/UDG2/5toSem/DeepLearning/Actividades/MTCombo.mp4")

# Propiedades del video
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))

# Guardar en la carpeta actual (donde está el notebook)
output_path = os.path.join(os.getcwd(), "segmentacion.mp4")
out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))

while True:
    ret, im0 = cap.read()
    if not ret:
        print("Procesamiento del video terminado.")
        break

    # Solo segmentación (sin tracking)
    results = model.predict(im0, verbose=False)

    # Dibujar únicamente las máscaras: sin cajas, etiquetas ni confianza
    im0 = results[0].plot(boxes=False, labels=False, conf=False, line_width=2)

    out.write(im0)

    cv2.imshow("segmentacion", im0)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

out.release()
cap.release()
cv2.destroyAllWindows()
print(f"Video guardado en: {output_path}")
```
